# The Message Hierarchy
### * Message
### * SystemMessage
### * UserMessage
### * AIMessage

```python
class Message:
    def __init__(self, role, content):
        self.role = role
        self.content = content

    def __str__(self):
        return f"{self.__class__.__name__}(role={self.role}, content={self.content})"

class SystemMessage(Message):
    def __init__(self, content):
        super().__init__("system", content)

class UserMessage(Message):
    def __init__(self, content):
        super().__init__("user", content)

class AIMessage(Message):
    def __init__(self, content):
        super().__init__("assistant", content)

class ToolMessage(Message):
    def __init__(self, content):
        super().__init__("tool", content)
```

# Message Object creation

```python
message1 = SystemMessage(content='You are a System Archiectect')
message2 = UserMessage(content='Create a design for the following application')
message3 = AIMessage(content="I have performed the following design")
message4 = ToolMessage(content="The result of tooling is this")

print(message1)
print(message2)
print(message3)
print(message4)
```

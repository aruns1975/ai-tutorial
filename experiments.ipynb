# First experiment for operator overloading in python

Matric class with operator overloaded

```python
from langchain_core.runnables import RunnableSequence


class Matrix:

    def __init__(self, data):
        self.data = data
        self.rows = len(data)
        self.cols = len(data[0])

    def _check_dimensions(self, other):
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError(
                "Matrices must have the same dimensions"
            )

    def __add__(self, other):
        self._check_dimensions(other)

        result = [
            [
                self.data[i][j] + other.data[i][j]
                for j in range(self.cols)
            ]
            for i in range(self.rows)
        ]

        return Matrix(result)

    def __sub__(self, other):
        self._check_dimensions(other)

        result = [
            [
                self.data[i][j] - other.data[i][j]
                for j in range(self.cols)
            ]
            for i in range(self.rows)
        ]

        return Matrix(result)

    def __mul__(self, other):
        self._check_dimensions(other)

        result = [
            [
                self.data[i][j] * other.data[i][j]
                for j in range(self.cols)
            ]
            for i in range(self.rows)
        ]

        return Matrix(result)

    def __truediv__(self, other):
        self._check_dimensions(other)

        result = [
            [
                self.data[i][j] / other.data[i][j]
                for j in range(self.cols)
            ]
            for i in range(self.rows)
        ]

        return Matrix(result)

    def __or__(self, other):
        self._check_dimensions(other)

        result = [
            [
                self.data[i][j] | other.data[i][j]
                for j in range(self.cols)
            ]
            for i in range(self.rows)
        ]

        return Matrix(result)

    def __repr__(self):
        return "\n".join(str(row) for row in self.data)
```

Testing the matrix class

```python
A = Matrix([
    [1, 2],
    [3, 4]
])

B = Matrix([
    [10, 20],
    [30, 40]
])

print(f"The result of addition is \n{A + B}")
print(f"The result of subtraction is \n{A - B}")
print(f"The result of multiplication is \n{A * B}")
print(f"The result of division is \n{A / B}")
print(f"The result of or is \n{A | B}")

```
